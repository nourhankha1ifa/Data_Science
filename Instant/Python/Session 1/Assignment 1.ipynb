# fruits = ("apple", "banana", "cherry", "mango", "orange")
* ## Prints the first and last fruit from the tuple.
* ## Insert "Yes" in the third place
* ## Print the the fourth one  

```python
fruits = ("apple", "banana", "cherry", "mango", "orange")

print(f"The First element is ({fruits[0]})\nthe Last element is ({fruits[-1]})")
```

# football_players = {"Alice", "Bob", "Charlie", "David"}

* ## Finds and prints the students who play only football but not basketball.
* ## Finds and prints the students who play only basketball but not football.
* ## Prints all students who play either football or basketball (or both).

```python
football_players = {"Alice", "Bob", "Charlie", "David"}
basketball_players = {"Charlie", "Eve", "Alice", "Frank"}
```

```python
both = football_players & basketball_players
both
```

```python
football = football_players - basketball_players
football
```

```python
basketball = basketball_players - football_players
basketball
```

```python
either = basketball_players | football_players
either
```

## Sets

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7}
```

```python
common = set_a & set_b
common
```

```python
a = set_a - set_b
a
```

```python
duplicates = [1,1,1,2,2,2,2,3,4,4,5,5,5,5,5]

set_dup = set(duplicates)

list_dup = list(set_dup)

print(list_dup)
```
