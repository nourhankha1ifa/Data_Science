# Python collection methods — full reference

Detailed reference for the built-in methods and functions used with **lists**, **tuples**, **sets**, and **dictionaries**, each with a short code example.

---

## 1. List (mutable, ordered, allows duplicates)

| Method | Description | Example |
|---|---|---|
| `append(x)` | Adds one item to the end | `fruits.append("mango")` |
| `extend(iterable)` | Adds all items from another iterable | `fruits.extend(["kiwi", "pear"])` |
| `insert(i, x)` | Inserts item at index `i` | `fruits.insert(1, "grape")` |
| `remove(x)` | Removes the first matching item | `fruits.remove("mango")` |
| `pop(i=-1)` | Removes and returns item at index (default: last) | `last = fruits.pop()` |
| `clear()` | Removes all items | `fruits.clear()` |
| `index(x)` | Returns index of first match | `fruits.index("apple")` |
| `count(x)` | Counts occurrences of `x` | `fruits.count("apple")` |
| `sort(key=None, reverse=False)` | Sorts the list in place | `fruits.sort(reverse=True)` |
| `reverse()` | Reverses the list in place | `fruits.reverse()` |
| `copy()` | Returns a shallow copy | `new_list = fruits.copy()` |

**Built-in functions that work on lists:**

```python
numbers = [4, 1, 7, 3]

len(numbers)        # 4
min(numbers)         # 1
max(numbers)         # 7
sum(numbers)         # 15
sorted(numbers)      # [1, 3, 4, 7] (returns new list, original unchanged)
list("abc")          # ['a', 'b', 'c']
```

---

## 2. Tuple (immutable, ordered, allows duplicates)

Tuples have very few methods because they cannot be changed after creation.

| Method | Description | Example |
|---|---|---|
| `count(x)` | Counts occurrences of `x` | `point.count(0)` |
| `index(x)` | Returns index of first match | `point.index(5)` |

**Built-in functions that work on tuples:**

```python
point = (5, 0, 3, 5)

len(point)           # 4
min(point)            # 0
max(point)            # 5
sum(point)            # 13
sorted(point)         # [0, 3, 5, 5] (returns a list)
tuple([1, 2, 3])      # (1, 2, 3)

# common companions
list(enumerate(point))       # [(0,5),(1,0),(2,3),(3,5)]
list(zip(point, point))      # pairs elements from both tuples
```

> Since tuples are immutable, "modifying" one really means creating a new tuple, e.g. `point + (9,)`.

## 3. Set (mutable, unordered, unique items only)

| Method | Description | Example |
|---|---|---|
| `add(x)` | Adds a single item | `colors.add("green")` |
| `update(iterable)` | Adds multiple items | `colors.update(["black", "white"])` |
| `remove(x)` | Removes item; raises error if missing | `colors.remove("red")` |
| `discard(x)` | Removes item; no error if missing | `colors.discard("purple")` |
| `pop()` | Removes and returns a random item | `item = colors.pop()` |
| `clear()` | Removes all items | `colors.clear()` |
| `union(other)` or `\|` | Combines two sets | `a.union(b)` / `a \| b` |
| `intersection(other)` or `&` | Common items only | `a.intersection(b)` / `a & b` |
| `difference(other)` or `-` | Items in `a` not in `b` | `a.difference(b)` / `a - b` |
| `symmetric_difference(other)` or `^` | Items in either, not both | `a ^ b` |
| `issubset(other)` | Checks if `a` is inside `b` | `a.issubset(b)` |
| `issuperset(other)` | Checks if `a` contains `b` | `a.issuperset(b)` |
| `isdisjoint(other)` | True if no common items | `a.isdisjoint(b)` |
| `copy()` | Returns a shallow copy | `b = a.copy()` |

**Built-in functions:**

```python
a = {1, 2, 3}
b = {2, 3, 4}

len(a)               # 3
set([1, 1, 2, 3])    # {1, 2, 3}  (removes duplicates)
frozenset(a)          # immutable version of a set
```

## 4. Dictionary (mutable, key–value pairs)

| Method | Description | Example |
|---|---|---|
| `get(key, default=None)` | Safe access, no error if key missing | `person.get("age", 0)` |
| `keys()` | Returns all keys | `person.keys()` |
| `values()` | Returns all values | `person.values()` |
| `items()` | Returns key–value pairs | `person.items()` |
| `update(other)` | Merges another dict or key/value pairs | `person.update({"age": 30})` |
| `pop(key, default=None)` | Removes key and returns its value | `person.pop("age")` |
| `popitem()` | Removes and returns the last inserted pair | `person.popitem()` |
| `setdefault(key, default)` | Returns value; sets it if key is missing | `person.setdefault("city", "Cairo")` |
| `clear()` | Removes all items | `person.clear()` |
| `copy()` | Returns a shallow copy | `p2 = person.copy()` |
| `fromkeys(keys, value)` | Creates a dict from a list of keys | `dict.fromkeys(["a","b"], 0)` |

**Built-in functions:**

```python
person = {"name": "Ali", "age": 25}

len(person)          # 2
dict(name="Sara")    # {'name': 'Sara'}

for key, value in person.items():
    print(key, value)
```

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
