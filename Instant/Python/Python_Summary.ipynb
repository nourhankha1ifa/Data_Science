## How to Solve Dictionary Questions

A dictionary stores information as **key-value pairs**.

For example, a student’s name can be the key and their grade the value:

```python
grades = {"Mona": 92, "Omar": 78, "Sara": 95}
```

When solving a dictionary problem, first identify what each **key** represents and what each **value** represents. Then choose the operation the question asks for.

| If the question asks you to… | Pattern | Example |
|---|---|---|
| Get a value from a known key | `dictionary[key]` | `grades["Mona"]` → `92` |
| Get a value safely when a key might be missing | `dictionary.get(key, default)` | `grades.get("Ali", 0)` → `0` |
| Add a new pair | `dictionary[key] = value` | `grades["Ali"] = 88` |
| Change a value | Assign to its existing key | `grades["Mona"] = 94` |
| Remove a known key | `dictionary.pop(key)` | `grades.pop("Omar")` |
| Visit every key and value | `for key, value in dictionary.items()` | Print each student and grade |
| Calculate a total or average | Use `.values()` with `sum()` and `len()` | `sum(grades.values()) / len(grades)` |
| Find the key with the largest value | `max(dictionary, key=dictionary.get)` | Returns the student with the highest grade |
| Find the key with the smallest value | `min(dictionary, key=dictionary.get)` | Returns the student with the lowest grade |
| Keep only matching pairs | Loop through `.items()` and check a condition | Find grades above the average |
| Check whether a key exists | `key in dictionary` | `if "Mona" in grades:` |
| Work with nested data | Access the inner dictionary using its outer key | `inventory["Laptop"]["price"]` |

### A reliable solving process

1. **Identify the dictionary’s structure.** Ask what the keys and values represent. Are the values numbers, text, or another dictionary?
2. **Find the requested result.** Does the question need a value, a key, a total, a filtered dictionary, or an updated dictionary?
3. **Choose the correct loop or method.** Use `.values()` for calculations on values, `.items()` when you need both keys and values, and `max(..., key=dictionary.get)` to find the key associated with the largest value.
4. **Check special cases.** Consider an empty dictionary, a missing key, or ties for the highest or lowest value.
5. **Print or return the result.** Use `print()` to display it; use `return` when writing a function that should give the result back to its caller.

### Common patterns

**Loop through a dictionary:**

```python
for name, grade in grades.items():
    print(f"{name}: {grade}")
```

**Calculate the average value:**

```python
average = sum(grades.values()) / len(grades)
```

**Find the key with the highest value:**

```python
highest_student = max(grades, key=grades.get)
highest_grade = grades.get(highest_student)
```

`max(grades)` compares the keys themselves.

`max(grades, key=grades.get)` compares the values associated with those keys and returns the key whose value is highest.

**Filter entries:**

```python
above_90 = {}

for name, grade in grades.items():
    if grade > 90:
        above_90[name] = grade
```

**Process nested dictionaries:**

```python
inventory = {
    "Laptop": {"price": 25000, "quantity": 5},
    "Phone": {"price": 15000, "quantity": 8}
}

for product, details in inventory.items():
    product_value = details["price"] * details["quantity"]
    print(product, product_value)
```

### 7.3 Attributes

- **Instance attribute:** unique per object, defined in `__init__` using `self`.
- **Class attribute:** shared by *all* instances, defined directly in the class body (outside methods).

```python
class Car:
    wheels = 4                       # class attribute (shared)
    def __init__(self, color, model):
        self.color = color           # instance attribute
        self.model = model

car1 = Car("Red", "Porsche")
car2 = Car("Blue", "BMW")
print(car1.color, car2.color)        # Red Blue (different)
print(Car.wheels, car1.wheels)       # 4 4 (same, shared)
```

### 7.4 Methods

- **Instance method:** takes `self`, works with that object’s data.
- **Class method:** `@classmethod`, takes `cls`, works with class-level data only.
- **Static method:** `@staticmethod`, takes neither `self` nor `cls` — just a helper grouped in the class.

```python
class Car:
    wheels = 4

    def __init__(self, color, model):
        self.color = color
        self.model = model

    def display_info(self):                    # instance method
        print(f"Car: {self.model}, Color: {self.color}")

    @classmethod
    def display_wheels(cls):                    # class method
        print(f"All cars have {cls.wheels} wheels.")

class Math:
    @staticmethod
    def add(a, b):                               # static method
        return a + b
print(Math.add(5, 3))    # 8
```

### 7.5 The Four Pillars of OOP

#### 🔒 Encapsulation

**Definition:** bundling data + methods that operate on it into one unit (a class). **Goal:** restrict direct access to protect data from unauthorized change.

| Access level | Naming convention | Meaning |
|---|---|---|
| Public | `self.name` | Can be accessed from anywhere. |
| Protected | `self._name` | Intended for use inside the class and its subclasses. Python does not enforce this; it is a convention. |
| Private | `self.__name` | Python changes the attribute name internally (name mangling), making direct access less convenient outside the class. |

**Getters/setters** provide controlled access & validation.

**Getter** returns a value
**Setter** changes it after checking that the new value is valid.

```python
class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number    # private
        self.__balance = balance                  # private

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive!")

    def withdraw(self, amount):
        if 0 < amount < self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient balance or invalid amount!")

    def get_balance(self):                          # getter
        return self.__balance

account = BankAccount("12345678", 5000)
account.deposit(2000)
account.get_balance()      # 7000
```

⚠️ A private attribute `__balance` is accessible outside only via name-mangling: `account._BankAccount__balance` (shows it’s a *convention*, not true security).

#### 🎭 Abstraction

**Definition:** hides complex implementation, shows only essential features.

**Goal:** simplify interaction (e.g. an ATM — you withdraw money without knowing the internal banking logic; a car’s steering wheel/pedals hide the engine mechanics).

Implemented in Python via **abstract base classes** (`abc` module) — forces subclasses to implement required methods.


```python
from abc import ABC, abstractmethod

class Shape(ABC):                     # abstract class
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width, self.height = width, height
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2
    def perimeter(self):
        return 2 * 3.14 * self.radius

rect = Rectangle(5, 10)
print(rect.area(), rect.perimeter())     # 50 30
```

⚠️ You **cannot instantiate** `Shape()` directly — it’s abstract. Every subclass **must** implement all `@abstractmethod`s or it also stays abstract.

#### 🧬 Inheritance

**Definition:** a class (child) inherits attributes/methods from another class (parent) → hierarchical structure.

**Goal:** code reuse + relationships between classes.

**Benefits:** code reusability, modularity, extensibility (override/extend parent methods), maintainability (parent changes auto-flow to children).

`super()` calls the parent class’s method (commonly `__init__`).

**Single inheritance** — One child inherits from one parent.

```python
class Animal:
    def __init__(self, name, color, age):
        self.name, self.color, self.age = name, color, age
    def speak(self):
        return "I make a sound."

class Dog(Animal):
    def speak(self):                 # overriding
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

dog = Dog("Buddy", "Golden", "11M")
print(dog.speak())    # Woof!
```

Using `super()` to extend `__init__` in the child:

```python
class Cat(Animal):
    def __init__(self, name, kind, color, age):
        super().__init__(name, kind, color)   # reuse parent init
        self.age = age
    def speak(self):
        return "Meow!"
```

**Multiple inheritance** — child inherits from more than one parent:

```python
class Engine:
    def start(self):
        print("Engine started")

class Radio:
    def play_music(self):
        print("Playing music")

class Car(Engine, Radio):    # inherits from BOTH
    def drive(self):
        print("Car is driving")

my_car = Car()
my_car.start()        # from Engine
my_car.play_music()   # from Radio
my_car.drive()        # own method
```

**Multilevel inheritance** — A class inherits from a child class. a chain (A → B → C):

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity
    def display_info(self):
        print(f"Brand: {self.brand}, Model: {self.model}, Battery: {self.battery_capacity} kWh")

my_car = ElectricCar("Tesla", "Model S", 100)
my_car.display_info()
```

#### 🔀 Polymorphism

**Definition:** one interface (method name), many forms — different objects respond differently to the same method call. **Goal:** flexibility & reusability.

- **Method overriding:** child redefines a method already in the parent (see `Dog`/`Cat` above).
- **Method overloading:** same method name, different parameters (Python doesn’t support true overloading like Java — usually simulated with default/`*args`).

```python
class Bird:
    def fly(self):
        return "I can fly."

class Sparrow(Bird):
    def fly(self):
        return "I fly at low altitudes."

class Eagle(Bird):
    def fly(self):
        return "I soar high in the sky."

class Penguin(Bird):
    def fly(self):
        return "I cannot fly, but I swim!"

birds = [Sparrow(), Eagle(), Penguin()]
for bird in birds:
    print(bird.fly())     # same call, different behavior per object
```

### 7.6 Class counters (shared class attribute)


```python
class Student:
    Nstudents = 0                       # class attribute
    def __init__(self, name, age, major):
        self.name, self.age, self.major = name, age, major
        Student.Nstudents += 1          # increments every time a new object is made

    def display_info(self):
        print(f"{self.name}, {self.age}, {self.major}, Count: {Student.Nstudents}")
```

### 8.1 Sales analysis (function returning multiple results)

```python
def analyze_sales(sales):
    total = sum(sales.values())
    average = total / len(sales)
    highest_name = max(sales, key=sales.get)
    above_average = {n: a for n, a in sales.items() if a > average}
    bonuses = {n: a * 0.10 for n, a in sales.items() if a > 15000}
    return total, average, highest_name, above_average, bonuses

sales = {"Ahmed": 12000, "Mona": 18000, "Omar": 9000, "Sara": 22000, "Ali": 15000}
total, average, highest, above_avg, bonuses = analyze_sales(sales)
```

### 8.2 Menu-driven program pattern (e.g. Employee Management)

Recurring structure:
**(1)** one function per operation → **(2)** `while True` menu loop → **(3)** `if/elif` to dispatch the user’s choice → **(4)** `break` to exit.

```python
employees = {
    "E001": {"name": "Ahmed", "department": "IT", "salary": 18000},
    "E002": {"name": "Mona", "department": "HR", "salary": 15000},
}

def display_all():
    for emp_id, details in employees.items():
        print(emp_id, details)

def average_salary():
    return sum(p["salary"] for p in employees.values()) / len(employees)

while True:
    print("1 Display  2 Average  3 Exit")
    try:
        choice = int(input("Choose: "))
    except ValueError:
        print("Enter a menu number")
        continue

    if choice == 1:
        display_all()
    elif choice == 2:
        print(f"Average salary: {average_salary():.2f}")
    elif choice == 3:
        break
    else:
        print("Choose 1-3")
```

4.  Put reusable logic in functions; `return` useful results.
5.  Check edge cases: empty data, missing keys, invalid values, exact thresholds.
6.  Keep object state changes **behind class methods** (encapsulation).
