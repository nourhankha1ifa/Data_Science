## 5. Dictionaries

A `dict` stores data as **key-value pairs**. Keys must be unique, and each key maps to a value.


### Dictionary - Q1

Question:
Create a dictionary containing information about a student:
name, age, department, and grade.
Then:
1. Print the student's name and department.
2. Update the grade.
3. Add a new key called email.
4. Print the complete dictionary.

```python
students = {
    "name": "Nourhan",
    "age": 22,
    "department": "Data Science",
    "grade": "B"
}

print(f"Student name: {students["name"]}\nDepartment: {students["department"]}\n")

students["grade"] = "A"

students["email"] = "nourhankhalifa186@gmail.com"

for key, value in students.items():
    print(f"{key}: {value}")
```

### Dictionary - Q2

Question: Given the following dictionary:

products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 10000,
    "Headphones": 3000
}

1. Print all product names.
2. Print all prices.
3. Calculate and print the average price.
4. Find and print the most expensive product.

```python
products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 10000,
    "Headphones": 3000
}

for product, price in products.items():
    print(f"{product}: {price}")

print(f"\nAverage price: {sum(products.values())/len(products)}")

print(f"Most expensive product: {max(products.keys())} -> {max(products.values())}")
```

### Dictionary - Q3

Question: Given the following dictionary:

student_grades = {
    "Ahmed": 85,
    "Mona": 92,
    "Omar": 78,
    "Sara": 95,
    "Ali": 88
}

1. Calculate the average grade.
2. Find the student with the highest grade.
3. Find the student with the lowest grade.
4. Print students whose grades are above the average.

```python
student_grades = {
    "Ahmed": 85,
    "Mona": 92,
    "Omar": 78,
    "Sara": 95,
    "Ali": 88
}

average = sum(student_grades.values())/len(student_grades.keys())

print(f"Average students grade: {average}")

print(f"Highest student: {max(student_grades, key=student_grades.get)} with {max(student_grades.values())}")

print(f"Lowest student: {min(student_grades, key=student_grades.get)} with {min(student_grades.values())}")

print("Students above the average: ")
for key, value in student_grades.items():
    if value > average :
        print(f"  {key}: {value}")
```

### Dictionary - Q4

Question: Given the following inventory:

inventory = {
    "Laptop": {"price": 25000, "quantity": 5},
    "Phone": {"price": 15000, "quantity": 8},
    "Tablet": {"price": 10000, "quantity": 3}
}

Calculate the total inventory value for each product.

Formula: price * quantity

Then calculate the total value of the entire inventory.

```python
inventory = {
    "Laptop": {"price": 25000, "quantity": 5},
    "Phone": {"price": 15000, "quantity": 8},
    "Tablet": {"price": 10000, "quantity": 3}
}

total_inventory = 0

for product, details in inventory.items():
    value = details["price"] * details["quantity"]
    print(f"{product}: {value}")

    total_inventory += value

print("Total inventory value:", total_inventory)
```

## 6. if, elif, else (Conditional Statements)

Conditional statements let your program make decisions based on whether something is `True` or `False`.


### If Condition - Q1

Question: Ask the user to enter a student's grade.

Print:
- \>= 90  -> Excellent
- \>= 80  -> Very Good
- \>= 70  -> Good
- \>= 50  -> Pass
- < 50   -> Fail

```python
grade = float(input("Enter student's grade: "))

if grade >= 90:
    print("Excellent")
elif grade >= 80:
    print("Very Good")
elif grade >= 70:
    print("Good")
elif grade >= 50:
    print("Pass")
else:
    print("Fail")
```

### If Condition - Q3

Question: A company calculates employee bonuses based on performance:

- Performance >= 90 -> 20% bonus
- Performance >= 80 -> 15% bonus
- Performance >= 70 -> 10% bonus
- Performance < 70  -> No bonus

Ask the user for salary and performance score.

Calculate the bonus and final salary.

```python
salary = float(input("Enter salary: "))
performance = float(input("Enter performance score: "))

if performance >= 90:
    bonus = salary * 0.20
elif performance >= 80:
    bonus = salary * 0.15
elif performance >= 70:
    bonus = salary * 0.10
else:
    bonus = 0

final_salary = salary + bonus

print("Bonus:", bonus)
print("Final salary:", final_salary)
```

### If Condition - Q4

Question: Create a shopping discount system.

Customer types:

Regular:
   - Purchase < 5000  -> 0%
   - Purchase >= 5000 -> 10%

VIP:
   - Purchase < 5000  -> 15%
   - Purchase >= 5000 -> 20%

Calculate the final amount after discount.

```python
purchase = float(input("Enter purchase amount: "))
customer_type = int(input("Enter customer type (Regular = 1 or VIP = 2): "))

if customer_type == 1:
    if purchase >= 5000:
        discount = 0.10
    else:
        discount = 0

elif customer_type == 2:
    if purchase >= 5000:
        discount = 0.20
    else:
        discount = 0.15

else:
    print("Invalid customer type")
    discount = 0

discount_amount = purchase * discount
final_amount = purchase - discount_amount

print("Discount:", discount_amount)
print("Final amount:", final_amount)
```

## 7. for Loop

A `for` loop is used to iterate over a sequence (list, tuple, string, dictionary, range, etc.).


### For Loop - Q1

Question:

Given: numbers = [12, 5, 8, 20, 3, 15, 7]

Using a for loop:
1. Print all numbers.
2. Calculate the sum.
3. Calculate the average.
4. Count numbers greater than 10.

```python
numbers = [12, 5, 8, 20, 3, 15, 7]

total = 0
count = 0

for n in numbers:
    print(n, end=" ")
    total += n
    if n > 10:
        count += 1

avg = total / len(numbers)

print(f"\nSum: {total}\nAverage: {avg}\nCount of numbers greater than 10: {count}")
```

### For Loop - Q3

Question:

Given: numbers = [12, 7, 25, 8, 30, 15, 4, 18]

Using a for loop:
1. Create a list of even numbers.
2. Create a list of odd numbers.
3. Find the largest number.
4. Calculate the sum of numbers greater than 10.

```python
numbers = [12, 7, 25, 8, 30, 15, 4, 18]

total = 0

even = []
odd = []

for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)

print(f"Even numbers: {even}")
print(f"Odd numbers: {odd}")
print(f"Largest number: {max(numbers)}")

for n in numbers:
    if n > 10:
        total += n

print(f"Sum of numbers greater than 10: {total}")
```

### For Loop - Q4

Question:

Given:

sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

Using a for loop:
1. Calculate total sales.
2. Calculate average sales.
3. Find the employee with the highest sales.
4. Print employees above the average.
5. Calculate a 10% bonus for employees with sales > 15000.

```python
sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

average = sum(sales.values()) / len(sales)

print(f"Total Sales: {sum(sales.values())}")
print(f"Average: {average}")
print(f"\nHighest Sales: {max(sales, key=sales.get)} with sales {max(sales.values())}")

print("\nEmployees above average: ")
for name, sale in sales.items():
    if sale > average:
        print(f"{name}: {sale}")

print("\n10% bonus for employees with sales > 15000:")
for name, sale in sales.items():
    if sale > 15000:
        bonus = (sale * 0.10) + sale
        print(f"{name}: {bonus}")
```

## 8. while Loop

A `while` loop keeps running as long as its condition is `True`. Be careful to avoid infinite loops!


### While Loop - Q1

Question: Ask the user to enter numbers repeatedly.

Stop when the user enters 0.

Then print:
- Sum of entered numbers
- Number of numbers entered
- Average

```python
total = 0
count = 0

while i != 0:
    i = int(input("Enter number or (0 to Exit): "))
    total += i
    count += 1

average = total / count

print(f"Sum: {total}\nCount: {count}\nAverage: {average}")
```
