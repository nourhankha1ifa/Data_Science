## 9. Functions

Functions let you group reusable blocks of code. Python supports several kinds of functions:
- Regular functions
- Functions with default arguments
- Functions with `*args` (variable number of positional arguments)
- Functions with `**kwargs` (variable number of keyword arguments)
- Lambda (anonymous) functions
- Recursive functions


### Functions - Q1

Question: Create a function called `get_grade` that takes a student's grade as an argument and returns:

- \>= 90  -> "Excellent"
- \>= 80  -> "Very Good"
- \>= 70  -> "Good"
- \>= 50  -> "Pass"
- < 50   -> "Fail"

Test the function with different grades.


```python
grade = int(input("Enter your grade: "))

def get_grade(grade):
    if grade >= 90:
        print("Excellent")
    elif grade >= 80:
        print("Very Good")
    elif grade >= 70:
        print("Good")
    elif grade >= 50:
        print("Pass")
    else:
        print("Fail")

get_grade(grade)
```

### Functions + Dictionary - Q2

Question: Create a function called `analyze_sales` that takes a dictionary containing employees and their sales.

The function should:
1. Calculate the total sales.
2. Calculate the average sales.
3. Find the employee with the highest sales.
4. Return all three results.

Use the following data:

sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

```python
sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

def analyze_sales(sales):
    Total = sum(sales.values())
    Average = sum(sales.values())/ len(sales.keys())
    Max = max(sales.values())

    return Total, Average, Max

analyze_sales(sales)
```

## 10. try / except (Error Handling)

`try` / `except` lets you handle errors gracefully instead of letting your program crash.

Common Types of Exceptions
- ValueError → An invalid value.
- TypeError → An inappropriate data type.
- NameError → An undefined variable.
- IndexError → An index outside the list's boundaries.
- KeyError → A key that does not exist in a Dictionary.
- FileNotFoundError → The file does not exist.
- ZeroDivisionError → Division by zero.

### Try / Except - Q1

Question: Ask the user to enter two numbers and calculate their sum and division.

Use try/except to handle:
1. If the user enters something that is not a number.
2. If the user tries to divide by zero.

The program should display an appropriate message for each error.

```python
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")

try:
    num1 = float(num1)
    num2 = float(num2)
    print(f"\nThe sum of {num1} and {num2} is {num1 + num2}")
except ValueError:
    print("Invalid input")

try:
    print(f"The division of {num1} and {num2} is {num1 / num2}")
except ZeroDivisionError:
    print("Cannot divide by zero")
```

### Try / Except - Q2

Question: Given the following dictionary:

products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 10000
}

- Ask the user to enter a product name and the quantity they want to buy.
- Calculate and print the total price.

Use try/except to handle:
1. If the product does not exist in the dictionary.
2. If the user enters an invalid quantity.

Example:
- Product: Laptop
- Quantity: 2
- Total Price: 50000

```python
products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 10000
}

print(products)
product = input("Enter a product name: ").title()
quantity = int(input("Enter the quantity: "))

if product in products:
    try:
      print(f"\nProduct: {product}")
      print(f"Quantity: {quantity}")
      print(f"Total price: {products[product] * quantity}")
    except ValueError:
        print("Invalid quantity")
else:
    print("Product does not exist")
```
