### 📝 Practice Questions — Data Types

1. Create four variables: an `int`, a `float`, a `str`, and a `bool`. Print each value along with its type.
2. Given `x = "3.14"`, convert it to a `float`, then convert that result to an `int`. Print both results.


```python
#1
a = 3
b = 3.5
c = "Three"
d = True

print(f"value: {a}", type(a))
print(f"value: {b}", type(b))
print(f"value: {c}", type(c))
print(f"value: {d}", type(d))
```

```python
#2
x = "3.14"

x = float(x)
y = int(x)

print(f"Float: {x}")
print(f"Int: {y}")
```

### 📝 Practice Questions — Lists

1. Create a list of 5 numbers. Sort it in **descending** order, then print the minimum and maximum values.
2. Start with a list of names. Append a new name, then remove one of the original names, and print the final list.


```python
#1
num = [1, 2, 3, 4, 5]

num.sort(reverse=True)

print(f"Min: {min(num)}\nMin: {max(num)}")
```

```python
#2
names = ["Mirelle", "Malak", "Haya", "Joyce", "Kenzy"]

names.append("Nourhan")
names.remove("Kenzy")

names
```

### 📝 Practice Questions — Sets

1. Given `set_a = {1, 2, 3, 4, 5}` and `set_b = {4, 5, 6, 7}`, find the elements that are common to both, and the elements that exist **only** in `set_a`.
2. Given a list with duplicate values, remove the duplicates using a `set` and print the result as a list.


```python
#1
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7}
```

```python
common = set_a & set_b
common
```

```python
only_a = set_a - set_b
only_a
```

```python
#2
duplicates = [1,1,1,2,2,3,4,4,5,5,5,5]

values = set(duplicates)
values
```

### 📝 Practice Questions — Dictionaries

1. Create a dictionary of 3 students with their grades. Loop through it and print each entry as `"name: grade"`.
2. Starting from an existing dictionary: add a new key-value pair, update an existing value, delete one key, then print the final dictionary.


```python
#1
students = {"Nourhan": "84",
            "Mariam": "90",
            "Ali": "75"}

for name, grade in students.items():
    print(f"{name}: {grade}")
```

```python
#2
#Add
students["Kareem"] = "93"

#Update
students["Mariam"] = "77"

#Remove
students.pop("Ali")

students
```
