Question: Given the following list of numbers:

numbers = [12, 7, 25, 8, 30, 15, 4, 18, 21, 10]

Write a program to:
1. Calculate the sum of all numbers.
2. Calculate the average.
3. Find the maximum number.
4. Find the minimum number.
5. Count how many numbers are greater than the average.
6. Create a new list containing only the even numbers.

````python
numbers = [12, 7, 25, 8, 30, 15, 4, 18, 21, 10]

avg = sum(numbers) / len(numbers)

# 1:4
print(f"Sum: {sum(numbers)}")
print(f"Average: {avg}")
print(f"Max: {max(numbers)}")
print(f"Min: {min(numbers)}")

# 5
count = 0

for n in numbers:
  if n > avg:
    count += 1

print(f"Numbers greater than Average: {count}")

#6
even = []

for num in numbers:
  if num % 2 == 0:
    even.append(num)

print(f"Even numbers: {sorted(even)}")
````

Question: Given the following dictionary:

employees = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

Write a program to:
1. Calculate the total salaries.
2. Calculate the average salary.
3. Find the employee with the highest salary.
4. Find the employee with the lowest salary.
5. Print employees whose salary is above the average.
6. Create a new dictionary containing only employees with salaries >= 15000

````python
employees = { "Ahmed": 12000,
              "Mona": 18000,
              "Omar": 9000,
              "Sara": 22000,
              "Ali": 15000
            }

avg_emp = sum(employees.values()) / len(employees)

# 1:4
print(f"Total salaries: {sum(employees.values())}")
print(f"Average salary: {avg_emp}")
print(f"Highest salary: {max(employees, key=employees.get)}")
print(f"Lowest salary: {min(employees, key=employees.get)}")

# 5
emp_above_avg = {}

for key, value in employees.items():
  if value > avg_emp:
    emp_above_avg[key] = value

print(f"Employees with salary above average: {emp_above_avg}")

# 6
salaries = {}

for k, v in employees.items():
  if v >= 15000:
    salaries[k] = v

print(f"Employees with salary >= 15000: {salaries}")
````

Question: Create a program that analyzes a list of student grades:

grades = [85, 92, 67, 78, 95, 88, 72, 60, 90, 55]

Using a for loop:
1. Calculate the average grade.
2. Count the number of students who passed. `(Passing grade = 50)`
3. Count the number of students who failed.
4. Find the highest grade.
5. Find the lowest grade.
6. Create a list containing grades above the average.

````python
grades = [85, 92, 67, 78, 95, 88, 72, 60, 90, 55]

# 1
avg = sum(grades) / len(grades)
print(f"Average grade: {avg}")

# 2:3
passed = []
failed = []

for grade in grades:
  if grade >= 50:
    passed.append(grade)
  else:
    failed.append(grade)

print(f"Passed: {len(passed)}")
print(f"Failed: {len(failed)}")

# 4:5
print(f"Highest grade: {max(grades)}")
print(f"Lowest grade: {min(grades)}")

# 6
above_avg = []

if grade > avg:
  above_avg.append(grade)

print(f"Grades above average: {above_avg}")
````

## Python Basics Practice - Q10

Question: Create a function called `calculate_discount`.

The function should take:
- price
- customer_type

Customer types:

1. Regular:
  ```
  price < 5000  -> 0% discount
  price >= 5000 -> 10% discount
  ```
2. VIP:
  ```
  price < 5000  -> 15% discount
  price >= 5000 -> 20% discount
  ```
The function should return:
- Discount amount
- Final price

Test the function with different customer types and prices.

````python
def calculate_discount(price, customer_type):
    if customer_type == "regular":
        if price < 5000:
            discount = 0
        else:
            discount = 0.10
    elif customer_type == "vip":
        if price < 5000:
            discount = 0.15
        else:
            discount = 0.20
    else:
        return "Invalid Customer Type"

    discount_amount = price * discount
    final_price = price - discount_amount

    return discount_amount, final_price


price = float(input("Enter amount: "))
customer_type = input("Enter Customer Type (Regular/VIP): ").lower()

discount_amount, final_price = calculate_discount(price, customer_type)

print(f"Discount amount: {discount_amount}")
print(f"Final price: {final_price}")
````

Question: Given the following dictionary:
```
products = {
    "Laptop": {"price": 25000, "quantity": 5},
    "Phone": {"price": 15000, "quantity": 8},
    "Tablet": {"price": 10000, "quantity": 3},
    "Headphones": {"price": 3000, "quantity": 10}
}
```
Write a program to:
1. Calculate the total value of each product.
   `Formula: price * quantity`
2. Calculate the total value of the entire inventory.
3. Find the product with the highest inventory value.
4. Create a new dictionary containing products
   whose quantity is less than 6.

Use a for loop to process the dictionary.

````python
products = {
    "Laptop": {"price": 25000, "quantity": 5},
    "Phone": {"price": 15000, "quantity": 8},
    "Tablet": {"price": 10000, "quantity": 3},
    "Headphones": {"price": 3000, "quantity": 10}
}

# 1
product_values = {}
for name, details in products.items():
  value = details["price"] * details["quantity"]
  product_values[name] = value
  print(f"Total value of {name}: {value}")

# 2
total_inventory = sum(product_values.values())
print(f"\nTotal Inventory: {total_inventory}")

# 3
highest_product = max(product_values, key=product_values.get)
highest_value = product_values[highest_product]
print(f"\nProduct with Highest Inventory Value: {highest_product} ({highest_value})")

# 4
low_quantity = {}
for name, details in products.items():
  if details["quantity"] < 6:
    low_quantity[name] = details

print(f"\nProducts with quantity less than 6: {low_quantity}")
````

## Python Basics Practice - Q12

Question: Create a simple ATM program.

Initial balance: balance = 10000

Display the following menu repeatedly:
1. Check Balance
2. Deposit
3. Withdraw
4. Exit

Requirements:
- Check Balance should display the current balance.
- Deposit should increase the balance.
- Withdraw should decrease the balance.
- Do not allow withdrawal if the amount is greater than the balance.
- Do not allow negative deposit or withdrawal amounts.
- The program should continue running until the user selects Exit.

Use a while loop and if conditions.

````python
print("   ATM program")
print("------------------")
print("1. Check Balance")
print("2. Deposit")
print("3. Withdraw")
print("4. Exit")

balance = 10000

while True:
  choice = int(input("\nEnter your desired operation: "))

  if choice == 1:
    print(f"Current Balance: {balance}")
  elif choice == 2:
    deposit = float(input("Deposit: "))
    if deposit < 0:
      print("Can not allow negative deposit")
    else:
      balance += deposit
  elif choice == 3:
    withdraw = float(input("Withdraw: "))
    if withdraw > balance:
      print("The amount is greater than the balance.")
    elif withdraw < 0:
      print("Can not allow negative Withdraw")
    else:
      balance -= withdraw
  elif choice == 4:
    print("\nSee You Soon!")
    break
  else:
    print("Invalid Input, try again!")
````

## Python Basics Practice - Q14

Question: Create a function called `analyze_sales`.

The function receives a dictionary containing employee names
and their sales:
```
sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}
```
The function should:
1. Calculate total sales.
2. Calculate average sales.
3. Find the employee with the highest sales.
4. Find the employee with the lowest sales.
5. Create a list of employees whose sales are above the average.
6. Calculate a 10% bonus for employees whose sales are greater than 15000.

Return the results from the function.

Use:
- Function
- Dictionary
- for loop
- if condition

````python
sales = {
    "Ahmed": 12000,
    "Mona": 18000,
    "Omar": 9000,
    "Sara": 22000,
    "Ali": 15000
}

def analyze_sales(sales):
  total_sales = sum(sales.values())
  avg_sales = total_sales / len(sales)
  highest_sales = max(sales, key=sales.get)
  lowest_sales = min(sales, key=sales.get)

  above_avg = {}
  for name, value in sales.items():
    if value > avg_sales:
      above_avg[name] = value

  bonus = {}
  for name, value in sales.items():
    if value > 15000:
      bonus[name] = round(value * 1.1, 2)

  return total_sales, avg_sales, highest_sales, lowest_sales, above_avg, bonus

total_sales, avg_sales, highest_sales, lowest_sales, above_avg, bonus = analyze_sales(sales)

print(f"Total sales: {total_sales}")
print(f"Average sales: {avg_sales}")
print(f"Highest sales: {highest_sales} ({sales[highest_sales]})")
print(f"Lowest sales: {lowest_sales} ({sales[lowest_sales]})")
print(f"Employees with sales above average: {above_avg}")
print(f"Employees with 10% bonus: {bonus}")
````

## Python Basics Practice - Q15

Question: Create a complete Employee Management System.

Use the following dictionary:

```
employees = {
    "E001": {
        "name": "Ahmed",
        "department": "IT",
        "salary": 18000
    },
    "E002": {
        "name": "Mona",
        "department": "HR",
        "salary": 15000
    },
    "E003": {
        "name": "Omar",
        "department": "IT",
        "salary": 22000
    },
    "E004": {
        "name": "Sara",
        "department": "Finance",
        "salary": 20000
    }
}
```

Create a menu using a while loop:
1. Display all employees
2. Search for an employee by ID
3. Calculate average salary
4. Find the employee with the highest salary
5. Display employees from a specific department
6. Add a new employee
7. Exit

Requirements:
- Use functions to organize the program.
- Use if/elif for menu choices.
- Use loops to search and process employees.
- Use try/except when necessary to handle invalid input.
- Make sure the program continues running until the user selects Exit.

The goal is to combine the following Python concepts:
- Variables
- Data Types
- Strings
- Lists
- Dictionaries
- If Conditions
- For Loops
- While Loops
- Functions
- Try/Except

````python
employees = {
    "E001": {
        "name": "Nourhan",
        "department": "IT",
        "salary": 22000
    },

    "E002": {
        "name": "Mohamed",
        "department": "HR",
        "salary": 15000
    },

    "E003": {
        "name": "Ahmed",
        "department": "HR",
        "salary": 18000
    },

    "E004": {
        "name": "Jasmine",
        "department": "Finance",
        "salary": 20000
    }
}


def display_employees():
    print("\nAll Employees:")
    print("----------------")
    for emp_id, details in employees.items():
        print(f"\nID: {emp_id}")
        for key, value in details.items():
            print(f"{key}: {value}")


def search_employee():
    emp_id = input("\nEnter employee ID: ").upper()

    if emp_id in employees:
        print(f"\nEmployee ID: {emp_id}")
        for key, value in employees[emp_id].items():
            print(f"{key}: {value}")
    else:
        print("Employee not found")


def average_salary():
    total_salary = 0
    for details in employees.values():
        total_salary += details["salary"]

    avg_salary = total_salary / len(employees)

    print(f"\nAverage Salary: {avg_salary:.2f}")


def highest_salary():
    highest = max(
        employees.values(),
        key=lambda employee: employee["salary"]
    )

    print(
        f"\nEmployee with highest salary: "
        f"{highest['name']} ({highest['salary']})"
    )


def department():
    dept = input("\nEnter department name: ").lower()

    found = False

    print(f"\nEmployees from {dept.title()} department:")

    for emp_id, details in employees.items():
        if details["department"].lower() == dept:
            found = True
            print(f"\nID: {emp_id}")
            for key, value in details.items():
                print(f"{key}: {value}")

    if not found:
        print("No employees found in this department.")


def add_employee():
    emp_id = input("\nEnter employee ID: ").upper()

    if emp_id in employees:
        print("Employee already exists")
    else:
        name = input("Enter employee name: ").capitalize()
        department = input("Enter employee department: ").strip()

        if len(department) == 2:
            department = department.upper()
        else:
            department = department.capitalize()

        try:
            salary = float(input("Enter employee salary: "))

            new_emp = {
                "name": name,
                "department": department,
                "salary": salary
            }

            employees[emp_id] = new_emp

            print("\nEmployee added successfully!")

        except ValueError:
            print("Invalid salary. Please enter a number.")


print("Employee Management System")
print("---------------------------")

while True:

    print("\n1. Display all employees")
    print("2. Search for an employee by ID")
    print("3. Calculate average salary")
    print("4. Find the employee with the highest salary")
    print("5. Display employees from a specific department")
    print("6. Add a new employee")
    print("7. Exit")

    try:
        choice = int(input("\nEnter a number to choose from the Menu: "))

        if choice == 1:
            display_employees()

        elif choice == 2:
            search_employee()

        elif choice == 3:
            average_salary()

        elif choice == 4:
            highest_salary()

        elif choice == 5:
            department()

        elif choice == 6:
            add_employee()

        elif choice == 7:
            print("\nSee You Soon!")
            break

        else:
            print("Invalid Input. Please choose from 1 to 7.")

    except ValueError:
        print("Invalid input. Please enter a number.")
````
